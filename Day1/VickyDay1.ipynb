```python
import numpy as np
import pandas as pd

# data loading
data = pd.read_csv("/root/.cache/kagglehub/datasets/patelris/crop-yield-prediction-dataset/versions/7/yield_df.csv")
df = pd.DataFrame(data)
print(df)
```

```python
# convert both fields into numpy array
yield_crops = df['hg/ha_yield'].to_numpy()
rainfall = df['average_rain_fall_mm_per_year'].to_numpy()

#ensure datasplit is random in nature for that u ensure that whatever indices u choose their order is random thats what this line does
# rng is short for random number generator on change in seed, sequence generated also changes, for same seed same sequence is produced
idx = np.random.default_rng(0).permutation(len(yield_crops))

#do a 80 20 split
cut = int(0.8 * len(yield_crops))
tr, te = idx[:cut], idx[cut:]

#print related metrics
print("Rainfall (Training Split) Metrics:")
print("Mean:", rainfall[tr].mean(),"Variance:", rainfall[tr].var(),"Standard Deviation:", rainfall[tr].std())
print("Rainfall (Testing Split) Metrics:")
print("Mean:", rainfall[te].mean(),"Variance:", rainfall[te].var(),"Standard Deviation:", rainfall[te].std())
print("Yield Crops (Training Split) Metrics:")
print("Mean:", yield_crops[tr].mean(),"Variance:", yield_crops[tr].var(),"Standard Deviation:", yield_crops[tr].std())
print("Yield Crops (Testing Split) Metrics:")
print("Mean:", yield_crops[te].mean(),"Variance:", yield_crops[te].var(),"Standard Deviation:", yield_crops[te].std())
```
